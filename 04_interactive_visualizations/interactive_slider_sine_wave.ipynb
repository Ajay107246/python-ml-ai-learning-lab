```python
# Interactive slider (change frequency live)
# This creates a slider that updates the sine wave dynamically.
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

x = np.linspace(0, 10, 500)
freq = 1

fig, ax = plt.subplots()
plt.subplots_adjust(bottom=0.25)

line, = ax.plot(x, np.sin(freq * x))
ax.set_title("Interactive Sine Wave")

# Slider
ax_freq = plt.axes([0.2, 0.1, 0.65, 0.03])
slider = Slider(ax_freq, 'Frequency', 0.1, 5.0, valinit=freq)

def update(val):
    line.set_ydata(np.sin(slider.val * x))
    fig.canvas.draw_idle()

slider.on_changed(update)

plt.show()

# Move the slider → the wave changes instantly.
```
