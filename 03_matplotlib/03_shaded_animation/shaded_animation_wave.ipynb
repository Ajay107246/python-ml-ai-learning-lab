```python
# Animate a sine wave using Matplotlib's FuncAnimation

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation

# Generate x-values
x = np.linspace(0, 10, 100)

# Create figure and axes
fig, ax = plt.subplots()

# Plot the initial sine wave
line, = ax.plot(x, np.sin(x))

# Update the wave for each animation frame
def update(frame):
    y = np.sin(x + frame / 10)
    line.set_ydata(y)
    return line,

# Create the animation
ani = FuncAnimation(
    fig,
    update,
    frames=200,
    interval=50
)

# Display the animation
plt.show()

###########################################################

# Shaded area under the curve
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 200)
y = np.sin(x)

plt.plot(x, y, color='blue')
plt.fill_between(x, y, alpha=0.3, color='skyblue')   # shading

plt.title("Sine Wave with Shaded Area")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
# Display the animation
plt.show()
```
