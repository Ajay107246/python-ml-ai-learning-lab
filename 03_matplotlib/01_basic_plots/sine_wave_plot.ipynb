```python
# draws a simple sine wave plot using NumPy for the math and Matplotlib for the graph.
import matplotlib.pyplot as plt
import numpy as np

# Generate 100 evenly spaced values between 0 and 10
x = np.linspace(0, 10, 100)

# Calculate the sine of each x value
y = np.sin(x)

#plt.plot(x, np.sin(x))

# Create the plot
plt.plot(x, y)

# Add title and axis labels
plt.title("Sine Wave")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Display the plot
plt.show()
```
