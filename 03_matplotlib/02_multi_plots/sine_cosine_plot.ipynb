```python
# Plot sine and cosine waves using NumPy and Matplotlib

import matplotlib.pyplot as plt
import numpy as np

# Generate 100 evenly spaced values between 0 and 10
x = np.linspace(0, 10, 100)

# Calculate sine and cosine values
sine_wave = np.sin(x)
cosine_wave = np.cos(x)

# Plot sine wave
plt.plot(
    x,
    sine_wave,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Sine"
)

# Plot cosine wave
plt.plot(
    x,
    cosine_wave,
    color="blue",
    linestyle="-",
    linewidth=2,
    label="Cosine"
)

# Add title and axis labels
plt.title("Sine and Cosine Waves")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Add gridlines
plt.grid(True, linestyle=":", linewidth=0.7, alpha=0.7)

# Display legend
plt.legend()

# Display the plot
plt.show()
```
