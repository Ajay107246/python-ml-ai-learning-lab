# 📊 Interactive Classification Threshold Demo
This notebook demonstrates how changing the classification threshold affects:
- Confusion Matrix
- Accuracy
- Precision
- Recall
- F1 Score

It refers from the behavior shown in Google's ML Crash Course (ROC & AUC section).

NOTE: https://developers.google.com/machine-learning/crash-course/classification/thresholding

```python
import numpy as np
# python -m pip install numpy pandas scikit-learn ipywidgets
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
import ipywidgets as widgets
from IPython.display import display, clear_output

# -----------------------------
# 1. Generate Synthetic Dataset
# -----------------------------
np.random.seed(42)

# Ground truth labels (25% positive)
y_true = np.array([1]*25 + [0]*75)

# Predicted probabilities
probs = np.concatenate([
    np.random.uniform(0.4, 0.95, 25),  # Actual positives
    np.random.uniform(0.05, 0.6, 75)   # Actual negatives
])

def display_metrics(threshold):
    """Compute confusion matrix + metrics for a given threshold."""

    # Convert probabilities to binary predictions
    y_pred = (probs >= threshold).astype(int)

    # Confusion matrix values
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

    # Metrics
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, zero_division=0)
    rec = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    # Confusion matrix table
    cm_df = pd.DataFrame({
        "Actual Positive": [f"TP: {tp}", f"FN: {fn}"],
        "Actual Negative": [f"FP: {fp}", f"TN: {tn}"]
    }, index=["Predicted Positive", "Predicted Negative"])

    # Metrics table
    metrics_df = pd.DataFrame({
        "Metric": ["Accuracy", "Precision", "Recall (TPR)", "F1-Score"],
        "Value": [f"{acc:.2f}", f"{prec:.2f}", f"{rec:.2f}", f"{f1:.2f}"]
    })

    # Display output
    clear_output(wait=True)
    print(f"Current Classification Threshold: {threshold:.2f}\n")
    print("--- Confusion Matrix ---")
    display(cm_df)

    print("\n--- Evaluation Metrics ---")
    display(metrics_df)

    # Helpful notes
    if threshold > 0.8:
        print("\nNote: High threshold increases Precision but decreases Recall.")
    elif threshold < 0.2:
        print("\nNote: Low threshold increases Recall but decreases Precision.")


# -----------------------------
# 2. Create Interactive Slider
# -----------------------------
threshold_slider = widgets.FloatSlider(
    value=0.5,
    min=0.0,
    max=1.0,
    step=0.05,
    description='Threshold:',
    continuous_update=True
)

# -----------------------------
# 3. Interactive Output
# -----------------------------
widgets.interactive(display_metrics, threshold=threshold_slider)
```
